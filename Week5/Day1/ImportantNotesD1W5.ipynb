```python
# Bag of words (BoW)
# “Bag of Words” (BoW) is a term used in NLP that refers to a way of extracting features from text for use in modeling, such as in machine learning algorithms.
# In the BoW model, a text (such as a sentence or a document) is represented as the bag (multiset) of its words, disregarding grammar and even word order but keeping multiplicity.

# In practical terms, the BoW model involves two primary steps:

# Vocabulary Creation: Identify all the unique words in the entire text dataset and create a vocabulary. Each unique word becomes a feature.

# Feature Vector Creation: For each document or piece of text, create a vector. The length of this vector is equal to the length of the vocabulary. The vector is filled with the count (frequency) of each word from the vocabulary in the document.

# To understand the concept of creating the vocabulary and frequency of the words, lets use Counter from python collections.
# Use the code from the previous exercises with a different text:
```

```python
# Counter is a good tool to exemplify the concept, but in terms of text analysis, many times we will have a collection of documents (a collection with strings as elements, where each string is a document).
# For those cases we have different models with different approuchs to extract feautures. Let’s undertand them:

# CounterVectorizer:
# Purpose: Converts text documents into vectors of word counts.
# Mechanism: It creates a vocabulary of all the unique words in the corpus and represents each document as a vector with counts of how often each word appears.
# Characteristic: It treats every word as equally important and does not account for the context in which a word appears. The order of words is also ignored.
# Use Case: Good for basic text classification tasks where word occurrence is important.
```

```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ['This is the first document.','This document is the second document.','And this is the third one.', 'This one is the 4th']

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)

print("Feature names:", vectorizer.get_feature_names_out())
print("Vectorized representation:\n", X.toarray())

# Notice that the matrix has 4 rows (each one representing one doc in the docs list), the leng of each row is the lenght of the document, and the order of the words is ignored.
```

```python
# TF-IDF (Term Frequency-Inverse Document Frequency):
# Purpose: Weighs the word frequencies by a measure of how unique they are to the document.
# Mechanism: It multiplies the term frequency (how often a word appears in a document) with the inverse document frequency (a measure of how rare or common a word is across all documents).
# Characteristic: It gives more weight to words that are unique to a particular document, helping to identify more important words and reduce the weight of common words.
# Use Case: Useful in search engines and information retrieval because it highlights specific words that are unique to a document.


# So let’s understand how it works with an example:

# corpus = 10.000 documents
# docs lenght = 100 words
# serach word = “food”


# Let’s say the TF (term frequency) will be 30 times in one document. The TF score will be calculated this way:
# how many times/doc_lenght
# 30/100 = .3

# Now let’s say the word “food” was found in 352 documents. The IDF score will be calculated this way:
# log(corpus/docs_found) = IDF
# log(10.000/352) = 1.45



# Finally, the TF-IDF is calculated this way:
# TF*IDF
# .3 * 1.45 = .44
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Step 1: Load the Documents
documents = [
    "I love reading books.",
    "Reading in the morning is refreshing.",
    "I love morning coffee."
]

# Step 2: Apply TF-IDF Vectorization
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

# Step 3: Analyze the Output
# Print the vocabulary (unique words)
print("Vocabulary:", vectorizer.get_feature_names_out())

# Print the TF-IDF values for each document
print("TF-IDF Matrix:")
print(tfidf_matrix.toarray())


# output:
# Vocabulary: ['books' 'coffee' 'in' 'is' 'love' 'morning' 'reading' 'refreshing' 'the']
# TF-IDF Matrix:
# [[0.68091856 0.         0.         0.         0.51785612 0.
#   0.51785612 0.         0.        ]
#  [0.         0.         0.44036207 0.44036207 0.         0.3349067
#   0.3349067  0.44036207 0.44036207]
#  [0.         0.68091856 0.         0.         0.51785612 0.51785612
#   0.         0.         0.        ]]
```

```python
# The TF-IDF Matrix is a 3x9 matrix where each row represents one of your documents and each column corresponds to one of the words in the vocabulary.
# The values in the matrix represent the TF-IDF score of each word in each document.
# A higher TF-IDF score indicates a word is more important or relevant in a specific document.


# Let’s use the first row to understand better:

# [0.68091856 0. 0. 0. 0.51785612 0. 0.51785612 0. 0. ]
# ‘books’ ‘coffeee’ ‘in’ ‘is’ ‘love’ ‘morning’ ‘reading’ ‘refreshing’ ‘the’

# We can notice that the most important word in the first document is ‘books’, which has the highest TF-IDF score.
```

```python
# Using the previous example, load this json file and analyse what is the most important word of each poem:

from google.colab import files
uploaded = files.upload()


import json
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('stopwords')

# Load JSON file
with open('famous_poems.json', 'r') as f:
    data = json.load(f)

# Clean and preprocess poems
stop_words = set(stopwords.words('english'))
processed_poems = []

for poem in data:
    tokens = word_tokenize(poem['text'])
    filtered_tokens = [word for word in tokens if word.lower() not in stop_words and word.isalpha()]
    processed_poems.append(' '.join(filtered_tokens))

# Apply TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(processed_poems)

# Get feature names (vocabulary)
features = vectorizer.get_feature_names_out()

# BONUS: Get the most important word (highest TF-IDF score) in each poem
for i, poem in enumerate(processed_poems):
    row = tfidf_matrix[i].toarray()[0]
    max_index = row.argmax()
    most_important_word = features[max_index]
    print(f"Poem {i+1}: Most important word → {most_important_word}")
```

```python
# Challenge : Classify Text as Spam or Not Spam
# Build a simple text classifier that categorizes messages as Spam or Not Spam using keyword-based classification.

# 1. Below are five text messages. Identify whether each message is Spam or Not Spam based on keywords like "win", "free", "offer", "meeting", and "project".

# 2. Write "Spam" or "Not Spam" next to each message based on your classification.

# 3. Explain your reasoning in one sentence.
```

```python
messages = [
    "Congratulations! You've won a free vacation to Bahamas!",
    "Let's schedule a project meeting for tomorrow.",
    "Limited time offer! Buy one get one free!",
    "Reminder: Your team meeting is at 3 PM today.",
    "You have been selected to win a brand new iPhone!"
]

# Step 1: Define spam keywords
spam_keywords = ['win', 'free', 'offer']

# Step 2: Classify each message and explain reasoning (a simple classification using keyword matching)

for i, msg in enumerate(messages):
    msg_lower = msg.lower()
    if any(keyword in msg_lower for keyword in spam_keywords):
        label = "Spam"
    else:
        label = "Not Spam"

    print(f"Message {i+1}: {label}")

    # Simple reasoning
    if label == "Spam":
        print("Reason: Contains common spam keyword(s) like 'win', 'free', or 'offer'.\n")
    else:
        print("Reason: Contains professional/neutral terms like 'meeting' or 'project', not typically associated with spam.\n")
```

```python
# Hands-On Implementation
# Let’s build a spam detection model using the Spam SMS Dataset.

# Step 1: Load and Preprocess the Data

import pandas as pd
from sklearn.model_selection import train_test_split

from google.colab import files
uploaded = files.upload()

# Load the dataset
df = pd.read_csv("spam.csv", encoding="latin-1")
df = df[["v1", "v2"]]  # Keep only the label and text columns
df.columns = ["label", "text"]

# Convert labels to binary (spam = 1, ham = 0)
df["label"] = df["label"].map({"spam": 1, "ham": 0})

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(df["text"], df["label"], test_size=0.2, random_state=42)
```

```python
# Step 2: Feature Extraction with TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

# Convert text to TF-IDF features
vectorizer = TfidfVectorizer(stop_words="english")
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
# Step 3: Train a Naive Bayes Classifier

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Train the model
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# Make predictions
y_pred = model.predict(X_test_tfidf)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
