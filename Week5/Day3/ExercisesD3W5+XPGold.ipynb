```python
# Step 2: Import the BERT Tokenizer and Initialize It

from transformers import BertTokenizer  # import the BERT tokenizer class from the transformers library

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')  # initialize the tokenizer using the 'bert-base-uncased' version
```

```python
# Choosing a Sample Sentence and Tokenize It

sentence = "This is a test sentence for tokenization."  # define a sample sentence to tokenize

tokens = tokenizer.tokenize(sentence)  # apply the tokenizer to break the sentence into subword tokens

print("Original Sentence:", sentence)
print("Tokenized Output:", tokens)     # prints the list of tokens created by the tokenizer
```

```python
# Encoding the Sentence with Special Tokens and Padding=
# using tokenizer.encode_plus() to:
# -Add special tokens that BERT needs (like [CLS] and [SEP])
# -Pad the sentence to a fixed length
# -Return the results as PyTorch tensors

encoded = tokenizer.encode_plus(
    sentence,                    # the input sentence
    add_special_tokens=True,     # add [CLS] at the start and [SEP] at the end
    padding="max_length",        # pad the sentence if it's shorter than max_length
    max_length=15,               # set a fixed length for the sequence
    return_tensors="pt"          # return the result as PyTorch tensors (pt = PyTorch)
)
```

```python
# Print the Encoded Output – Token IDs and Corresponding Tokens=
# We look at the actual encoded output:
# Token IDs → Each token (word or subword) is mapped to a unique number from BERT’s vocabulary.
# Tokens → Human-readable version of what BERT sees after tokenization and special token insertion.

print("Token IDs:", encoded['input_ids'])  # print the list of numerical IDs for each token
print("Attention Mask:", encoded['attention_mask'])  # shows which tokens are real (1) and which are padding (0)

# To view the actual tokens corresponding to the token IDs
tokens_with_specials = tokenizer.convert_ids_to_tokens(encoded['input_ids'][0])  # convert IDs back to readable tokens
print("Tokens with Special Tokens and Padding:", tokens_with_specials)
```

```python
# Summary Notes:
# [CLS]: Always position 0 — classification starting point
# [SEP]: Marks sentence boundary (always required)
# [PAD]: Fills unused slots when max_length is larger than sentence
# Attention Mask: Highlights what’s real vs. what’s padding
# WordPiece tokens like ##ization are used for unknown or complex words
```

```python
# 1. Choose a knowledge source, such as a Wikipedia article or a collection of text documents.

# These are short text passages that our RAG system will use to find answers.

documents = [
    "Paris is the capital of France. It is known for the Eiffel Tower.",
    "The Great Wall of China is one of the most famous landmarks in the world.",
    "The Pacific Ocean is the largest and deepest ocean on Earth.",
    "Python is a popular programming language for data science and web development.",
    "The human heart pumps blood through the body and is a vital organ."
]

# Our system will search through these documents to find relevant information for a given question.
```

```python
# 2. Use a pre-trained BERT model to generate embeddings for the knowledge source and store them in a vector database (e.g., FAISS).

from transformers import AutoTokenizer, AutoModel  # BERT tokenizer and model
import torch

# load tokenizer and model (bert-base-uncased)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

# function to generate an embedding for a single text
def get_embedding(text):
    # tokenize and convert to tensor
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)

    # pass through BERT to get output embeddings
    with torch.no_grad():
        outputs = model(**inputs)

    # extract the [CLS] token embedding (sentence-level representation)
    return outputs.last_hidden_state[:, 0, :].squeeze(0)

# Embeddings are numerical vectors that capture the meaning of text
# We use the [CLS] token output from BERT as the document-level embedding
# These embeddings allow us to compare text based on meaning, not just keywords
# Later, we'll compare the user question embedding to these document embeddings
# FAISS is a fast vector search tool we’ll use to find the most similar documents
# This helps us build the "retrieval" part of the RAG system
```

```python
# 3. Implement a retriever that takes a user question as input, generates its embedding using BERT, and retrieves the most relevant documents from the vector database
# (converting a user question into an embedding and retrieve similar documents using FAISS)

import faiss
import numpy as np

# First, prepare and store all document embeddings in a FAISS index
doc_embeddings = []

for doc in documents:
    emb = get_embedding(doc)  # get BERT embedding for each document
    doc_embeddings.append(emb.numpy())  # convert PyTorch tensor to NumPy array

doc_embeddings = np.stack(doc_embeddings)  # stack all vectors into a matrix

# Create a FAISS index for fast similarity search using L2 (Euclidean distance)
dimension = doc_embeddings.shape[1]  # number of values in each vector
index = faiss.IndexFlatL2(dimension)  # create a simple flat index
index.add(doc_embeddings)  # store all document vectors in FAISS

# Storing all document embeddings in a FAISS index for fast retrieval
# FAISS lets us quickly search which document vector is most similar to the question vector
# Each document is converted into a vector using BERT and stored in the index
# Later, we will do the same for the user question and compare it to these vectors
```

```python
# 4. Use a pre-trained GPT model as the generator. Feed the retrieved documents and the user question to the generator to produce an answer.

from transformers import pipeline

# Load a text-to-text generation model (FLAN-T5 is lightweight and effective)
generator = pipeline("text2text-generation", model="google/flan-t5-base")

# Function to generate an answer from the question + retrieved document(s)
def generate_answer(question, retrieved_docs):
    # Combine context and question into a prompt
    context = " ".join(retrieved_docs)
    prompt = f"Context: {context}\n\nQuestion: {question}\n\nAnswer:"

    # Use the generator to produce an answer
    response = generator(prompt, max_length=100, do_sample=False)
    return response[0]['generated_text']

# we Load a pre-trained GPT-style model (we'll use FLAN-T5 here for simplicity) to generate answers
# The generator takes a prompt that includes context (retrieved document) + the question
# FLAN-T5 is a smaller model that works well for Q&A tasks and runs faster than GPT-3
# The generator returns a human-like answer using the context we provided
```

```python
# 5. Test the RAG system with various questions and evaluate its performance in terms of accuracy and relevance of the generated answers

# Testing the full RAG system (asking a question, retrieving documents, and generating an answer)

# Function to retrieve the most relevant documents using FAISS
def retrieve_document(question, top_k=1):
    # Generate embedding for the question using BERT
    question_embedding = get_embedding(question).numpy().reshape(1, -1)

    # Search in FAISS index for the top_k most similar documents
    distances, indices = index.search(question_embedding, top_k)

    # Return the top matching documents
    return [documents[i] for i in indices[0]]

# Example question
question = "What is the capital of France?"

# Step 1: Retrieving documents
retrieved = retrieve_document(question)

# Step 2: Generating answer using GPT-style model
answer = generate_answer(question, retrieved)

# Step 3: Displaying results
print("Question:", question)
print("Retrieved Document:", retrieved[0])
print("Generated Answer:", answer)

# Testing the full RAG pipeline: retrieval + generation
# Step 1: Convert the question into an embedding and retrieve the most relevant document using FAISS
# Step 2: Pass the retrieved document and question into the generator to get the final answer
# Step 3: Print the question, the matching document, and the generated answer
# You can try different questions to test how accurate and relevant the answers are
```

```python
# --- Improved RAG Version (Using Sentence-BERT + Cosine Similarity) ---

from sentence_transformers import SentenceTransformer
import numpy as np
import faiss

# Load Sentence-BERT model (better for semantic similarity)
sbert_model = SentenceTransformer("all-MiniLM-L6-v2")  # lightweight and effective

# Step 1: Create new document embeddings using Sentence-BERT
sbert_doc_embeddings = sbert_model.encode(documents, convert_to_numpy=True)

# Step 2: Normalize embeddings (for cosine similarity search via inner product)
sbert_doc_embeddings = sbert_doc_embeddings / np.linalg.norm(sbert_doc_embeddings, axis=1, keepdims=True)

# Step 3: Create a new FAISS index using inner product (cosine similarity)
dimension = sbert_doc_embeddings.shape[1]
sbert_index = faiss.IndexFlatIP(dimension)
sbert_index.add(sbert_doc_embeddings)
```

```python
# New Retriever Function (for SBERT version):

def sbert_retrieve_documents(question, top_k=1):
    # Get SBERT embedding for the question
    question_embedding = sbert_model.encode([question], convert_to_numpy=True)
    question_embedding = question_embedding / np.linalg.norm(question_embedding, axis=1, keepdims=True)

    # Search in SBERT-based FAISS index
    distances, indices = sbert_index.search(question_embedding, top_k)
    return [documents[i] for i in indices[0]]
```

```python
# Testing the Improved SBERT-Based RAG Pipeline

# (Testing the improved RAG system using SBERT for retrieval)
question = "What is the capital of France?"

retrieved_sbert_docs = sbert_retrieve_documents(question)
improved_answer = generate_answer(question, retrieved_sbert_docs)

print("\n--- Improved SBERT Version ---")
print("Question:", question)
print("Retrieved Document:", retrieved_sbert_docs[0])
print("Generated Answer:", improved_answer)
```
