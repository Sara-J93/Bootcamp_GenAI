Exercise 1 : Small Quizz
1- What is the key difference between traditional machine learning and deep learning? Traditional machine learning needs humans to choose important data features, while deep learning figures them out on its own using neural networks. This makes deep learning better for complex (unstructured) data like images, speech, and text.
2- How do artificial neural networks (ANNs) mimic the human brain? ANNs work like a simplified version of the human brain. They have neurons (nodes) that are connected by weights (like brain synapses). These neurons process information, learn patterns, and improve over time—just like how our brains learn from experience.
3- Why does deep learning perform better on large datasets compared to traditional machine learning? Deep learning gets better as it sees more data because its neural networks learn complex patterns. Traditional machine learning struggles with too much data since it relies on human-selected features, while deep learning automatically finds the best features from large datasets.
4- What are some challenges of deep learning, and how can they be addressed?
a-Needs a lot of data → Solution: we could use data augmentation or pre-trained models.
b-Computationally expensive → we could use GPUs/TPUs or cloud computing.
c-Hard to understand (black box problem) → we could use Explainable AI (XAI) tools.
d-Overfitting (memorizing instead of learning) → we could use dropout, regularization, or more data.
5- What is feature engineering, and why is it not needed in deep learning? Feature engineering is the process of manually selecting important data features for a machine learning model. In deep learning, this step is not needed because neural networks automatically learn the best features from raw data, making the process more efficient, especially for complex data like images and text.
6- What role do hidden layers play in a deep learning model? Hidden layers process and transform data step by step, learning increasingly complex patterns.Early layers detect simple patterns (e.g., edges in an image).
Middle layers recognize structures (e.g., eyes, wheels).
Deeper layers understand high-level concepts (e.g., a face, a car).
More hidden layers = better learning but more computing power needed.
7- In an artificial neural network (ANN), what is the function of an activation function? An activation function decides whether a neuron should be "activated" (turned on) or not. It helps the network learn complex patterns by adding non-linearity, meaning the model can handle real-world problems instead of just making simple calculations.
Exemples: ReLU → Most common, keeps positive values and sets negatives to zero.
Sigmoid → Converts values to a range between 0 and 1 (good for probabilities).
Tanh → Similar to Sigmoid but outputs between -1 and 1.

```python
# Exercise 3 : Building a Simple Neural Network with TensorFlow/Keras

# Step1 We import libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import mnist

# Step2 We import the dataset (x is features y is label)
(x_train, y_train), (x_test, y_test)= keras.datasets.mnist.load_data()

# Step3 Normalize and One-Hot Encode
x_train = x_train / 255.0 #(normalization)
x_test = x_test / 255.0 #(normalization)
print(x_train.shape)
print(x_test)
```

```python
# y_train = tf.keras.utils.to_categorical(y_train, num_classes=10) #(Tp do one-hot encoding)
y_train = tf.keras.utils.to_categorical(y_train, num_classes=10)
y_test = tf.keras.utils.to_categorical(y_test, num_classes=10)

print(y_train.shape)
print(y_test.shape)
```

```python
# Step4 Model architectur

My_model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),                          # Flatten the input
    keras.layers.Dense(units=128, activation='relu'),                    # Hidden layer 1 : A layer with 256 neurons and the ReLU activation function.
    keras.layers.Dense(units=10, activation='softmax')                   # Output layer :  Outputs 10 probabilities (one for each class(10 classes in Mnist)).
    ])

print(My_model.summary())
```

```python
# Step 5: Compile the Model
My_model.compile(optimizer='adam',
                 loss='sparse_categorical_crossentropy',
                 metrics=['accuracy'])
print(My_model.compile())

# Optimizer: Adam is a popular optimizer that adjusts the weights to minimize the loss.
# Loss Function: Categorical Crossentropy measures how well the model’s predictions match the true labels.
# Metrics: Accuracy is used to evaluate the model’s performance during training.
```

```python
My_model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
import numpy as np

y_train = np.argmax(y_train, axis=1)  # Convert one-hot encoding to integer labels
y_test = np.argmax(y_test, axis=1)
```

```python
My_model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
# Step 6: Train the Model
history = My_model.fit(
    x_train, y_train,  # Training data
    epochs=10,          # Train for 10 epochs
    validation_data=(x_test, y_test)  # We use test data for validation
)
print(history)

# Epochs: The number of times the model sees the entire dataset.
# Steps per Epoch: The number of batches processed in one epoch.
# Validation Data: Used to evaluate the model’s performance during training
```

```python
# Step 7: Make Predictions
Make_predictions = My_model.predict(x_test)
print(Make_predictions)
print(Make_predictions.shape)

# The model outputs a probability distribution for each image.
# For example, [0.1, 0.7, 0.05, …] means the model is 70% confident the image belongs to class 1.
```

```python
# This is just to make it clearer and To show actual digits
predicted_labels = Make_predictions.argmax(axis=1)
print(predicted_labels)
```

```python
# Step 8: Evaluate the Model’s performance on the test data

test_loss, test_accuracy = My_model.evaluate(x_test, y_test)
print(f"Test Accuracy: {test_accuracy:.2f}")

# The model’s accuracy on the test set is 0.98
```
