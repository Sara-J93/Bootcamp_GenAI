```python
# Exercise 2: Artificial Neural Networks (ANNs)

# 1. Draw (or use a diagram tool) to create a simple artificial neural network (ANN) with:

# An input layer (3 neurons)
# One hidden layer (4 neurons)
# An output layer (2 neurons)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Define the ANN Structure (model)
# Input layer (3 neurons), hidden layer (4 neurons), output layer (2 neurons)
model = Sequential([
    Dense(4, activation='relu', input_shape=(3,)),  #  Creates a hidden layer with 4 neurons and ReLU activation.
    Dense(2, activation='softmax')  #  Creates an output layer with 2 neurons for classification
])
```

```python
# Next we should compile the Model:

# Before training, we need to compile the model by specifying:
# Loss function (measures error)
# Optimizer (updates weights)
# Metrics (how we evaluate the model)

model.compile(optimizer='adam',   # Adam Optimizer helps adjust weights efficiently
              loss='categorical_crossentropy', # Categorical Crossentropy is great for classification tasks
              metrics=['accuracy']) # Accuracy helps track performance.
```

```python
model.summary()  #to display the num of layers, neurons per layer, and total parameters in the model.
```

```python
# OPTIONAL: Now we can Train the Model, I did it just for self training.
import numpy as np

# To generate random data (3 features per input)
X_train = np.random.rand(100, 3)  # 100 training examples
y_train = np.random.randint(0, 2, (100, 2))  # 100 labels (binary classification)

# To train the model
model.fit(X_train, y_train, epochs=5, batch_size=5)
```

3. Write a short description (3-5 sentences) explaining how information flows through the network.

An Artificial Neural Network (ANN) processes information in three steps. First, the input layer receives raw data (numbers, images, or text...) and passes it to the hidden layer. Next, in the hidden layer, each neuron applies a mathematical transformation using weights and biases. The activation function (ReLU in the hidden layer) decides which neurons should "fire" and send their values to the next layer. Finally, the output layer generates the final prediction, where the softmax activation function converts values into probabilities (e.g. whether an image is of a cat or a dog). The network learns by adjusting weights and biases through multiple training cycles to minimize errors.

```python
# Exercise 3: Creating the Dataset and Visualizing the Data
# 1. Import the necessary libraries: numpy, matplotlib.pyplot, and mean_squared_error from sklearn.metrics.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
```

```python
# 2. Create a set of 20 points using the formula ( y = -x^2 ),
#  adding normally distributed noise with a mean of 0 and a standard deviation of 0.05 using np.random.normal.

np.random.seed(0)  # For consistent results every time we run
x = np.arange(-1, 1, 0.1)  # Generate 20 points from -1 to 1
y = -x**2 + np.random.normal(0, 0.05, len(x))  # Adds random noise (mean = 0, std = 0.05)
```

```python
# to make sure i have 20 points
x = np.arange(-1, 1, 0.1)
print(x)
print("Total points:", len(x))
```

```python
# 3. Plot the generated points using a scatter plot to visualize the dataset with noise.
plt.figure(figsize=(6, 4))  # Set figure size
plt.scatter(x, y, color='blue', label="Noisy Data")  # Plot points
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Dataset with Noise (y = -x² + noise)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 4. Split the dataset into a training set (first 12 points) and a test set (last 8 points).
x_train = x[:12] # First 12 points for training
y_train = y[:12] # First 12 points for training
x_test = x[12:]  # last 8 points for testing
y_test = y[12:]  # last 8 points for testing
```

```python
print("Training Data:")
print("x_train:", x_train)
print("y_train:", y_train)
print("\nTesting Data:")
print("x_test:", x_test)
print("y_test:", y_test)
```

```python
# Exercise 4: Fitting Polynomial Models of Different Degrees
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial.polynomial import Polynomial
```

```python
# 1. Define a function polynomial_fit that takes a degree as input and returns a polynomial that best fits the training data.

def polynomial_fit(degree):
    """Fits a polynomial of a given degree to the training data."""
    coefficients = np.polyfit(x_train, y_train, degree)  # Find best coefficients
    poly = np.poly1d(coefficients)  # Create a polynomial equation
    return poly
```

```python
# 2. Define another function plot_polyfit that plots the training set, test set, and the fitted polynomial curve for a given degree.

def plot_polyfit(degree):
    """Plots the training data, test data, and fitted polynomial curve."""
    poly = polynomial_fit(degree)  # Get polynomial equation

# Generate smooth x-values for plotting
    x_smooth = np.linspace(min(x), max(x), 100)
    y_smooth = poly(x_smooth)  # Compute y-values for the curve
# Plot training data
    plt.scatter(x_train, y_train, color="blue", label="Training Data")
# Plot test data
    plt.scatter(x_test, y_test, color="red", label="Test Data")
# Plot polynomial curve
    plt.plot(x_smooth, y_smooth, color="green", label=f"Degree {degree} Fit")

    plt.xlabel("x values")
    plt.ylabel("y values")
    plt.title(f"Polynomial Fit of Degree {degree}")
    plt.legend()
    plt.grid(True)
```

```python
# 3. Use plot_polyfit to visualize the fit for polynomials of degree 1, 7, and 11.

plot_polyfit(1)   # A straight line (for degree 1)
plt.show()

plot_polyfit(7)  # Plot for degree 7 (A more flexible curve)
plt.show()

plot_polyfit(11)  # Plot for degree 11 (A very wiggly curve (probably overfitting))
plt.show()
```

```python
# 4. Observe how the model behaves on the training and test sets as the degree of the polynomial increases.
```

- As the degree increases:

1. Too low (degree 1) → Doesn’t fit the data well (underfitting).
2. Medium (degree 7) → Fits well without memorizing too much (good fit).
3. Too high (degree 11) → Memorizes the training data too much but fails on new data (overfitting).

```python
# Exercise 5: Cross-Validation to Find the Optimal Degree

# IN EXERCISE 3 I already generated the dataset and split it into training and test sets

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# 1- Loop through polynomial degrees from 1 to 11

degrees = range(1, 12)  # Degrees from 1 to 11
results = []  # Store (degree, RMSE for training, RMSE for test)

for degree in degrees:
    # Convert x values into polynomial features
    poly = PolynomialFeatures(degree)
    x_train_poly = poly.fit_transform(x_train.reshape(-1, 1))
    x_test_poly = poly.transform(x_test.reshape(-1, 1))

    # Train a Linear Regression model on the polynomial features
    model = LinearRegression()
    model.fit(x_train_poly, y_train)

    # Predict on training and test sets
    y_train_pred = model.predict(x_train_poly)
    y_test_pred = model.predict(x_test_poly)

# 2- Calculate RMSE (error score)
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

# 3- Store results
    results.append((degree, rmse_train, rmse_test))

    print(f"Degree {degree}: Training RMSE = {rmse_train:.4f}, Test RMSE = {rmse_test:.4f}")
```

```python
results = np.array(results)  # Convert list to NumPy array for easy plotting
```

```python
# 4. Plot the RMSE values for both the training and test sets against the polynomial degree using a logarithmic scale for the y-axis.

plt.figure(figsize=(8, 5))
plt.plot(results[:, 0], results[:, 1], marker='o', label="Training RMSE", color='blue')
plt.plot(results[:, 0], results[:, 2], marker='s', label="Test RMSE", color='red')

plt.xlabel("Polynomial Degree")
plt.ylabel("RMSE (Log Scale)")
plt.yscale("log")  # Use logarithmic scale for better visualization
plt.title("RMSE vs Polynomial Degree")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 5. Identify the degree that minimizes the RMSE on the test set and confirm that it corresponds to the true underlying model (( y = -x^2 )).

best_degree = results[np.argmin(results[:, 2]), 0]  # Find degree with lowest test RMSE
print(f"Best polynomial degree based on test RMSE: {int(best_degree)}")
```
