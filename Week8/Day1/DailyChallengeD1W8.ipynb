#Step 1: Craft a Prompt That Controls Output Tone, Format, and Length


Prompt:
"Rewrite the following internal policy update into a friendly and clear tone for employees. Use bullet points to organize the information. Paraphrase the text without directly quoting it. Keep the total word count under 75 words.

Text:
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

#Step 2: Evaluate the Output


AI Output to Evaluate:
- Always use the company-approved secure VPN when accessing internal systems remotely.

- Avoid using unsecured connections or personal devices without proper security.

- This helps protect our confidential data and ensures safe communication for everyone.

**Relevance**:	Does the output stick to the message of the source content?
- Yes – It keeps the core idea: secure access, VPN only, no unsecured/personal devices.

**Clarity**:	Is the message easier to understand for non-technical staff?
- Yes – It’s written in clear, simple language.

**Structure**:	Is the information formatted into clear bullet points?
- Yes – 3 clear bullet points.

**Tone**	Is the language appropriate for internal employee communication?
- Yes – Friendly and informative without being too casual or too formal.

**Length**	Does it stay under 75 words?
- Yes – Total is about 40–45 words.

**Factual Accuracy**:	Did the model make up or remove any critical security details?
- Yes – Nothing important was removed or invented. It paraphrased accurately.

#Step 3: Detect and Mitigate Hallucinations

**Revised Prompt**:
"Rewrite the following internal policy update into a friendly and clear tone for employees. Use bullet points to organize the information. Paraphrase the content without directly quoting it. Keep the total word count under 75 words. Only paraphrase the content in the input text. Do not add new recommendations or mention technologies not present in the input."

This works because:
- I still control tone, format, paraphrasing, and length
- I now clearly say "no extra recommendations or technologies"
- This helps the AI stick to the original policy without improvising

#Step 4: Paraphrasing Deep Dive

**Prompt**:
"Paraphrase the following policy for a junior intern audience. Use plain and simple language. Avoid corporate or legal jargon. Write no more than 4 short bullet points. Keep the tone supportive and informative.

**Text**:
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

#Step 5: Quote Extraction Variant

**Prompt**:

"Extract one direct quote from the following text that best captures the core security policy. Do not paraphrase or rewrite it.

**Text**:

Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."

##1. In what kind of internal communication would quoting be more appropriate than paraphrasing?
- When writing official policy documents,
- In legal notices,
- In formal training when employees need to know the exact rule,
- Or when highlighting a new or updated rule word-for-word to avoid misunderstandings.

##2. When might quoting pose a risk?

- When the quote is too technical, formal, or full of jargon → employees might not understand it.
- When the audience is new, like interns or non-technical staff, quoting might confuse instead of help.
- It might also sound too cold or robotic if the goal is to make employees feel comfortable.
