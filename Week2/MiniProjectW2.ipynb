```python
# What is the difference between New York and California in terms of sales and profit?
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Grouped Bar Chart to compare the total sales and profit between New York and California.)
# Sample data (replace with your actual data)
states = ["New York", "California"]
sales = [data[data['State'] == "New York"]['Sales'].sum(), data[data['State'] == "California"]['Sales'].sum()]
profit = [data[data['State'] == "New York"]['Profit'].sum(), data[data['State'] == "California"]['Profit'].sum()]

x = np.arange(len(states))  # Position of the states
width = 0.4  # Width of bars

plt.figure(figsize=(8, 5))

# Plot bars
plt.bar(x - width/2, sales, width, label="Sales", color="blue")
plt.bar(x + width/2, profit, width, label="Profit", color="green")

plt.xlabel("State")
plt.ylabel("Amount")
plt.title("Sales & Profit Comparison: New York vs California")
plt.xticks(x, states)  # Set state names on x-axis

# Add legend
plt.legend()

# Show plot
plt.show()
```

```python
# Are there any differences among states in profitability?

state_summary = data.groupby('State')[['Sales', 'Profit']].sum()  # To group by state and calculate total Sales & Profit

# This calculates Profitability (%)
state_summary['Profitability (%)'] = (state_summary['Profit'] / state_summary['Sales']) * 100
```

```python
# To sort states by profitability
state_summary = state_summary.sort_values(by='Profitability (%)', ascending=False)
```

```python
# Plot the profitability by state
plt.figure(figsize=(12, 6))
plt.bar(state_summary.index, state_summary['Profitability (%)'], color='pink', edgecolor='purple')
plt.xlabel('State')
plt.ylabel('Profitability (%)')
plt.title('Profitability by State')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
print("Top 5 Most Profitable States:\n", state_summary.head())
print("\nBottom 5 Least Profitable States:\n", state_summary.tail())
```

```python
# Determine if 20% of the customers contribute to 80% of the profit

customer_profit = data.groupby('Customer Name')['Profit'].sum() # To group data by customer and calculate total profit

customer_profit = customer_profit.sort_values(ascending=False) # To sort customers by profit (in descending order)


customer_profit.head()
```

```python
# To calculate the cumulative sum of profit
customer_profit_cumulative = customer_profit.cumsum()

# Then to calculate the total profit
total_profit = customer_profit.sum()

# 80% profit threshold
profit_threshold = total_profit * 0.8

# Top customers contributing to 80% of the profit
top_customers = customer_profit_cumulative[customer_profit_cumulative <= profit_threshold].index

#  The top 20% of customers contributing to 80% of profit:
print(f"Top customers contributing to 80% of profit:\n{customer_profit.loc[top_customers]}")
```

```python
# Pareto chart (Cumulative Profit)
plt.figure(figsize=(10, 6))
plt.plot(customer_profit_cumulative / total_profit * 100, marker='o', color='blue', label='Cumulative Profit (%)')
plt.axhline(80, color='red', linestyle='--', label="80% Profit Threshold")
plt.xlabel('Customers')
plt.ylabel('Cumulative Profit (%)')
plt.title('Pareto Principle: Top Customers Contributing to 80% of Profit')
plt.xticks(range(0, len(customer_profit), len(customer_profit)//5), rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# What are the Top 20 cities by Sales ? Then by Profit ?
city_summary = data.groupby('City')[['Sales', 'Profit']].sum() # To group data by city and calculate total Sales & Profit

# Top 20 cities by Sales
top_20_sales = city_summary.sort_values(by='Sales', ascending=False).head(20)

# Top 20 cities by Profit
top_20_profit = city_summary.sort_values(by='Profit', ascending=False).head(20)

# Profitability for each city
city_summary['Profitability (%)'] = (city_summary['Profit'] / city_summary['Sales']) * 100

# Profitability for the Top 20 Cities (sorted by Sales)
top_20_sales_profitability = city_summary.loc[top_20_sales.index]
```

```python
# Plot the Cumulative curve in Sales by Customers. Can we apply Pareto principle to customers and Sales ?
# Yes we can apply Pareto principle to customers and Sales
```

```python
# To group by customer and calculate total Sales
customer_sales = data.groupby('Customer Name')['Sales'].sum()

# To sort customers by total Sales in descending order
customer_sales_sorted = customer_sales.sort_values(ascending=False)

# To calculate cumulative sum of Sales
customer_sales_cumulative = customer_sales_sorted.cumsum()

# To calculate the total sales
total_sales = customer_sales.sum()

# To normalize cumulative sales to percentage of total sales
customer_sales_cumulative_percent = (customer_sales_cumulative / total_sales) * 100

print("Cumulative Sales by Customers:\n", customer_sales_cumulative_percent)
```

```python
# Plot the cumulative sales curve
plt.figure(figsize=(10, 6))
plt.plot(customer_sales_cumulative_percent, marker='o', color='blue', label='Cumulative Sales (%)')
plt.axhline(80, color='red', linestyle='--', label="80% Sales Threshold")
plt.xlabel('Customers')
plt.ylabel('Cumulative Sales (%)')
plt.title('Cumulative Sales Curve by Customers')
plt.legend()
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

# the top 20% of customers contributing to 80% of the sales
top_20_percent_customers = customer_sales_cumulative_percent[customer_sales_cumulative_percent <= 80].index
print("Top 20% of Customers Contributing to 80% of Sales:\n", customer_sales.loc[top_20_percent_customers])
```
