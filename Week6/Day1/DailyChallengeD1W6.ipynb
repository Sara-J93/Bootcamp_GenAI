```python
# 1. Understanding BERT and XLM-RoBERTa
```

```python
# Importing the tokenizer classes for BERT and XLM-RoBERTa from the Hugging Face Transformers library
from transformers import BertTokenizer, XLMRobertaTokenizer
```

```python
# Loading the pre-trained BERT tokenizer from the 'bert-base-uncased' model
# This is initializing a tokenizer trained on lowercase English text
bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
```

```python
# Loading the pre-trained XLM-RoBERTa tokenizer from the 'xlm-roberta-base' model
# This is initializing a tokenizer trained on multilingual data (100+ languages)
xlmr_tokenizer = XLMRobertaTokenizer.from_pretrained("xlm-roberta-base")
```

```python
# 2. Tokenizing Text

# Defining a sample sentence to tokenize
sample_text = "It's truly inspiring how the bootcamp uploads class content without ever reading it — a bold commitment to surprise-based learning."
```

```python
# Tokenizing the sample text using BERT tokenizer

# Using encode_plus to get input_ids and attention_mask
bert_tokens = bert_tokenizer.encode_plus(
    sample_text,                       # Supplying the sentence to be tokenized
    add_special_tokens=True,          # Adding [CLS] and [SEP] automatically
    max_length=32,                    # Limiting/padding the sentence to 32 tokens
    padding='max_length',             # Padding with zeros if sentence is shorter than 32 tokens
    truncation=True,                  # Truncating if the sentence is longer than 32 tokens
    return_attention_mask=True,       # Including the attention mask to indicate real vs padded tokens
    return_tensors='pt'               # Returning as PyTorch tensors (we can use 'tf' for TensorFlow)
)

# encode_plus to get:
# input_ids: The numeric ID of each token, including special tokens like [CLS] and [SEP]
# attention_mask:	A binary mask (1 = real token, 0 = padding token), so the model knows what to ignore
```

```python
# Printing the BERT input IDs
print("BERT input_ids:", bert_tokens['input_ids'])

# Printing the BERT attention mask
print("BERT attention_mask:", bert_tokens['attention_mask'])

# Decoding input IDs back into readable tokens
print("Decoded tokens:", bert_tokenizer.decode(bert_tokens['input_ids'][0]))
```

```python
# Tokenizing the same sentence using XLM-RoBERTa tokenizer
xlmr_tokens = xlmr_tokenizer.encode_plus(
    sample_text,
    add_special_tokens=True,          # Adding <s> and </s> automatically
    max_length=32,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)
```

```python
# Printing the XLM-RoBERTa input IDs
print("XLM-RoBERTa input_ids:", xlmr_tokens['input_ids'])

# Printing the XLM-RoBERTa attention mask
print("XLM-RoBERTa attention_mask:", xlmr_tokens['attention_mask'])

# Decoding the tokens back to human-readable form
print("Decoded tokens:", xlmr_tokenizer.decode(xlmr_tokens['input_ids'][0]))
```

# Tokenizing Text (notes for me):

- input_ids:	The numerical form of the text for the model
- attention_mask:	A mask that tells the model what to focus on (1 = real word, 0 = padding)
- decode():	Converts token IDs back to text for us to read
- add_special_tokens:	Adds [CLS], [SEP] (BERT) or <s>, </s> (XLM-RoBERTa) automatically
- padding='max_length':	Pads to the length we set (here: 32)
- truncation=True:	Cuts off text if it’s too long
- return_attention_mask:	Tells model which tokens are padding
- return_tensors='pt':	Returns PyTorch-ready format
- So the code we write doesn't really change — but the tokenizer's internal logic and output does, based on which model you're using.

```python
# 5. Creating Cross-Validation Folds

# Here We’re going to split our dataset into multiple training/validation subsets, so we can:
# - Evaluate our model more reliably
# - Avoid overfitting
# - Make better use of the available data
# We’ll use Stratified K-Fold, which ensures each fold has a balanced distribution of labels (important for classification problems like ours with 3 classes: 0, 1, 2).
```

```python
# Importing StratifiedKFold from scikit-learn
from sklearn.model_selection import StratifiedKFold

# (Bringing in the K-Fold splitter that makes sure each fold keeps similar label proportions, which helps our evaluation be more fair and realistic)
```

```python
# Creating a new column in the DataFrame to hold the fold index
df['kfold'] = -1

# This will store the fold number each row belongs to (0 to 4 if we use 5 folds)
```

```python
# Shuffling the dataset before splitting
df = df.sample(frac=1).reset_index(drop=True)

# here we:
#  Shuffle the rows randomly (good practice before splitting into folds)
# Resetting the index so the row order looks clean after shuffling
```

```python
# Defining the Stratified K-Fold object (5 folds)
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# here we are:
# Telling scikit-learn "Split this data into 5 folds, and keep label distribution balanced."
# Setting random_state=42 for reproducibility (same split every time)
```

```python
# Creating the folds by assigning fold numbers
for fold, (train_idx, val_idx) in enumerate(kf.split(X=df, y=df['label'])):
    # Assigning the fold number to the validation set in that fold
    df.loc[val_idx, 'kfold'] = fold

# here we are:
# Looping through each fold split
# Assigning the validation rows for that fold a value in the kfold column
# After this, every row will have a number in kfold column from 0 to 4
```

```python
# Previewing the new dataset structure with fold assignments
df.head()
```

```python
# Checking how many samples are in each fold
print(df['kfold'].value_counts())
```

summary of 5 (for me)
- we Created kfold column	To store which fold each row belongs to
- Shuffled the dataset	To ensure random mixing before splitting
- Used StratifiedKFold	To balance label distribution in each fold
- Assigned fold numbers	So we can later pick training/validation splits easily

```python
#  Storing Folds in Lists (ING-style)

# Creating empty lists to store training and validation indexes
train_idx_list = []
val_idx_list = []

# Looping through folds again and storing index lists
for fold, (train_idx, val_idx) in enumerate(kf.split(X=df, y=df['label'])):
    train_idx_list.append(train_idx)
    val_idx_list.append(val_idx)
```
