```python
# 1- Setting up the environment inputs for our Emergency Dispatch Agent (Environment: What is the agent "seeing" or "hearing")

# Adding incoming call as text (or speech-to-text)
incoming_call_text = "My mom is having chest pain and trouble breathing!"

# Adding location data (from GPS or phone tower)
caller_location = {"lat": 48.8566, "lon": 2.3522}  # Paris example

# Adding patient symptom info (extracted from user input)
patient_symptoms = ["chest pain", "difficulty breathing"]

# incoming_call_text = the main message we’re receiving from the caller
# caller_location = GPS coordinates we're reading to know where to send help
# patient_symptoms = a list we’re filling with what the caller describes
```

```python
# 2- Defining the Tools

# Creating a mock symptom checker tool
def checking_symptoms(symptoms):
    # Running triage logic
    if "chest pain" in symptoms and "difficulty breathing" in symptoms:
        return "high"
    elif "headache" in symptoms:
        return "medium"
    else:
        return "low"

# Creating a mock ambulance scheduling tool
def scheduling_ambulance(location, urgency_level):
    # Triggering ambulance dispatch simulation
    return f"Ambulance dispatched to {location} for {urgency_level} urgency."

# checking_symptoms = this tool is deciding how serious the situation is based on the symptoms
# scheduling_ambulance = this tool is sending help if the problem is urgent enough
```

```python
# 3. Defining the State

#  Initializing the agent's state dictionary
agent_state = {
    "caller_id": "caller_001",                          # Tracking who's calling
    "location": caller_location,                        # Saving where the emergency is
    "symptoms": patient_symptoms,                       # Storing the reported symptoms
    "triage_level": None,                               # Keeping the urgency score
    "dispatch_status": None                             # Recording whether help was sent
}

# caller_id = we're assigning a unique ID to keep track of this call
# location = pulling location from earlier so the agent knows where to send help
# symptoms = keeping the symptoms list for decision-making
# triage_level = we will store "high", "medium", or "low" here after triage
# dispatch_status = this will show if an ambulance was sent or not
```

```python
# 4. Making Decisions and Taking Action

# Step 1: Running triage tool on symptoms
agent_state["triage_level"] = checking_symptoms(agent_state["symptoms"])

# Step 2: Taking action based on triage level
if agent_state["triage_level"] == "high":
    agent_state["dispatch_status"] = scheduling_ambulance(agent_state["location"], "high")
else:
    agent_state["dispatch_status"] = "Monitoring. No immediate dispatch needed."

# Step 1: checking_symptoms = we’re using the triage tool to find out how urgent this is
# Step 2: if it’s HIGH urgency, we’re calling the ambulance tool to send help
# Otherwise, we update the state to just monitor the situation
```

```python
# Final Agent State (for logging or debug)

# Printing the agent's full memory/state
for key, value in agent_state.items():
    print(f"{key}: {value}")
```

# 5- Choosing the Agent Type and Justifying It

-Chosen Agent Type: Hybrid Agent

Deliberative Layer (the “planner” part):
- It's reasoning about symptoms using checking_symptoms()
- It remembers previous inputs like location, symptoms, and triage status
- It makes decisions by analyzing context and planning what action to take

Reactive Layer (the “reflex” part):
- If urgency is high → immediately dispatching ambulance
- Instant response to life-threatening symptoms = real-time reaction

Summary:
# best agent for this case is Hybrid because:
- It stores state (caller info, symptom history, dispatch logs)
- It plans based on symptom severity
- It reacts instantly to emergencies

# Comparing Agent Types: Hybrid vs Deliberative vs Reactive

# Deliberative agent:
- Thinks before acting
- Uses internal models and multi-step planning
- Can simulate different outcomes
- Is slower but more thoughtful

So if we used a Deliberative-only agent, it would:
1. Not react instantly to symptoms.
2. Simulate multiple scenarios (e.g., "what if it's a panic attack vs heart attack?")
3. Possibly request more info from the caller.
4. Delay dispatch until it’s 100% sure based on structured reasoning.

# Reactive agent:
- Doesn’t remember anything from earlier (no state)
- Responds immediately to current input
- Uses if-then rules or simple logic
- Fast, but not smart or adaptive

If we used a Reactive Emergency Dispatch Agent:
1. It would read the current symptoms only (no history or caller ID).
2. It would not store any state (can’t track if help was already sent).
3. It would respond instantly — like a hardcoded triage bot.
4. Every input would be treated as if it’s the **first** time.


To sum up
- Reactive agent is fast, but forgetful and shallow
- Deliberative agent is smart and thorough, but slow.
It's safer for non-urgent, complex decisions (e.g., hospital referrals).
- Hybrid agent is better for emergencies where quick action is life-saving.
It combines the best of both — fast reactions + thoughtful planning


#What would happen if your agent did not maintain state?

Without state:
- The agent would forget the caller’s location and symptoms after each response
- It might ask the same questions again and again
- It wouldn’t know if it already dispatched help or not
- It couldn’t handle multi-step conversations like:
-- Caller: “She’s 82 and has chest pain”
-- Agent: “Where are you?”
-- Caller: “123 Main Street”
- Without state, it forgets the chest pain part and treats this like a new call
