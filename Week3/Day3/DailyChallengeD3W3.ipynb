```python
import pandas as pd

# Assuming your file is a CSV file
df = pd.read_csv('Crashes.csv', encoding='ISO-8859-1')

# Display the first few rows of the DataFrame
df.head()
```

```python
# Clean and preprocess the data, addressing missing values and categorizing data as needed.

#missing values
df.isnull().sum()
# dropping missing values
df = df.dropna()
print(df)
```

```python
# Convert dates and other relevant fields to appropriate formats.
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df.dtypes
```

```python
# 2. Exploratory Data Analysis:
# Use Pandas to explore basic statistics such as the number of crashes, fatalities, and survival rates.

df.describe()
```

```python
print(df.columns)
```

```python
# Calculate the number of crashes and fatalities
num_crashes = df['Flight #'].nunique()
num_fatalities = df['Fatalities'].sum()
```

```python
# Analyze the frequency of crashes over time to identify any trends.

# Plot frequency of crashes over time
df['year'] = df['Date'].dt.year
df.groupby('year').size().plot(kind='line', title='Crashes over Time')
```

```python
# Calculate survival rate

df['survivors'] = df['Aboard'] - df['Fatalities']
df['survival_rate'] = df['survivors'] / df['Aboard']
print(df['survival_rate'])
```

```python
# 3. Statistical Analysis:
# Apply SciPy to analyze the distribution of fatalities and survival rates. Calculate key statistics like mean, median, and standard deviation.

from scipy import stats

#mean, median, and std of fatalities
mean_fatalities = df['Fatalities'].mean()
median_fatalities = df['Fatalities'].median()
std_fatalities = df['Fatalities'].std()
print(f"Mean Fatalities: {mean_fatalities}")
print(f"Median Fatalities: {median_fatalities}")
print(f"Standard Deviation of Fatalities: {std_fatalities}")
```

```python
# Conduct a hypothesis test (e.g., comparing the average number of fatalities in different decades or regions).
# I'm comparing the average number of fatalities in different regions

Location1 = df[df['Location'] == 'Bazaar, Kansas']['Fatalities']
Location2 = df[df['Location'] == 'Atlantic City, New Jersey']['Fatalities']

t_stat, p_value = stats.ttest_ind(Location1, Location2)

print(f"T-statistic: {t_stat}, P-value: {p_value}")
```

```python
# comparing the average number of fatalities in different decades

# First I should extract the year from the 'Date' column
df['year'] = df['Date'].dt.year

# Filter data by two decades
decade1 = df[df['year'] < 1980]['Fatalities']
decade2 = df[df['year'] >= 1980]['Fatalities']

# Perform a t-test to compare fatalities between the two decades
t_stat, p_value = stats.ttest_ind(decade1, decade2)

# Output the results
print(f"T-statistic: {t_stat}, P-value: {p_value}")
```
